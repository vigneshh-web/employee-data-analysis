```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("clean_data.csv")

df.head()
```

```python
avg_salary = df.groupby("department")["salary"].mean()

avg_salary.plot(kind="bar", color="steelblue")

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")

plt.tight_layout()

plt.savefig("01_salary_by_department.png")
plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.hist(df["salary"], bins=10, color="steelblue", edgecolor="black")

plt.title("Distribution of Employee Salaries")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.savefig("02_salary_histogram.png")
plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.scatter(
    df["experience_years"],
    df["salary"],
    color="steelblue"
)

plt.title("Salary vs Experience")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary (₹)")

plt.tight_layout()
plt.savefig("03_salary_vs_experience.png")
plt.show()
```

```python

salary_trend = df.groupby("experience_years")["salary"].mean()

plt.figure(figsize=(8, 5))

plt.plot(salary_trend.index, salary_trend.values,
         color="steelblue", marker="o", label="Average Salary")

plt.title("Average Salary by Experience")
plt.xlabel("Experience (Years)")
plt.ylabel("Average Salary")
plt.legend()

plt.tight_layout()
plt.savefig("04_salary_trend.png")
plt.show()
```

```python

plt.figure(figsize=(9, 5))

df.boxplot(column="salary", by="department")

plt.title("Salary Distribution by Department")
plt.suptitle("")
plt.xlabel("Department")
plt.ylabel("Salary")

plt.tight_layout()
plt.savefig("05_salary_boxplot.png")
plt.show()
```

The goal of this project is to analyse employee data to understand salary patterns, differences between departments, and the relationship between employee experience and salary.

2. Dataset

3. Key Visualisations

Bar chart: Average salary by department.

Histogram: Distribution of employee salaries.

Scatter plot: Relationship between salary and experience.

4. Key Insights

The charts help compare average salaries across departments, understand salary distribution, and explore how experience relates to salary.

5. Recommendations

Review salary differences across departments.
